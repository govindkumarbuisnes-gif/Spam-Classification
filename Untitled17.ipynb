```python
from nltk.corpus import stopwords
import string
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def transform_text(text):
  text = text.lower()
  text = nltk.word_tokenize(text)
  y = []
  for i in text:
    if i.isalnum():
      y.append(i)
  text = y[:]
  y.clear()
  for i in text:
    if i not in stopwords.words('english') and i not in string.punctuation:
      y.append(i)
  t=y[:]
  y.clear()
  for i in t:
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
df["new-col"] = df["text"].apply(transform_text)
```

```python
tfidf = TfidfTransformer()

x_counts = cv.fit_transform(df['new-col'])
x = tfidf.fit_transform(x_counts).toarray()
```

```python
y = df['Target'].values
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size = 0.2,random_state=2)

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

```

```python
mnb = MultinomialNB()
mnb.fit(x_train,y_train)
y_pred2 = mnb.predict(x_test)
print(accuracy_score(y_test,y_pred2))
```

```python
def predict_spam_probability(message):

    # Preprocess
    transformed = transform_text(message)

    # Vectorize
    vector_input = cv.transform([transformed])

    # TF-IDF
    tfidf_input = tfidf.transform(vector_input)

    # Get probabilities
    prob = mnb.predict_proba(tfidf_input)[0]

    return {
        "Ham Probability": round(prob[0], 2),
        "Spam Probability": round(prob[1], 2)
    }
```

```python
msg3 = "Hey, are we meeting tomorrow for the project discussion?"
predict_spam_probability(msg3)
```

```python
import gradio as gr
import base64
import os
import glob

def classify_message(message):
    probabilities = predict_spam_probability(message)
    if probabilities["Spam Probability"] > probabilities["Ham Probability"]:
        classification = "SPAM"
    else:
        classification = "HAM"
    scores = {
        "Ham": float(probabilities["Ham Probability"]),
        "Spam": float(probabilities["Spam Probability"])
    }
    return classification, scores

# Automatically scan for any uploaded images in /content/
supported_formats = ["*.jpg", "*.jpeg", "*.png", "*.webp"]
found_images = []
for fmt in supported_formats:
    found_images.extend(glob.glob(f"/content/{fmt}"))

bg_css = "background-color: #f0f2f5;"
image_used = "None"

if found_images:
    # Use the first image found in the folder
    selected_image = found_images[0]
    image_used = os.path.basename(selected_image)
    try:
        with open(selected_image, "rb") as image_file:
            encoded_string = base64.b64encode(image_file.read()).decode()
        bg_css = f"background-image: url('data:image/jpeg;base64,{encoded_string}'); background-size: cover; background-position: center; background-repeat: no-repeat;"
        print(f"Success! Using '{image_used}' as your background.")
    except Exception as e:
        print(f"Error loading image {image_used}: {e}")
else:
    print("No image files (.jpg, .png, .jpeg) found in /content/. Please upload one or keep using the default color.")

custom_css = f"""
.gradio-container {{
    {bg_css} !important;
}}
"""

interface = gr.Interface(
    fn=classify_message,
    inputs=gr.Textbox(lines=2, label="Input Message"),
    outputs=[
        gr.Textbox(label="Result"),
        gr.Label(label="Confidence")
    ]
)

interface.launch(theme="ocean", css=custom_css, share=True)
```
